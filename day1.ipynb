```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 5, figsize=(15, 3))

# 1. Rectangle
image = np.zeros((400, 400), dtype=np.uint8)
image[ 0:250, 80:320] = 255
axes[0].imshow(image, cmap="gray")
axes[0].set_title("Rectangle")

# 2. Horizontal line
image = np.zeros((400, 400), dtype=np.uint8)
image[195:205, 50:350] = 255
axes[1].imshow(image, cmap="gray")
axes[1].set_title("Horizontal")

# 3. Vertical line
image = np.zeros((400, 400), dtype=np.uint8)
image[50:350, 195:205] = 255
axes[2].imshow(image, cmap="gray")
axes[2].set_title("Vertical")

# 4. Cross
image = np.zeros((400, 400), dtype=np.uint8)
image[195:205, 50:350] = 255
image[50:350, 195:205] = 255
axes[3].imshow(image, cmap="gray")
axes[3].set_title("Cross")

# 5. Border
image = np.zeros((400, 400), dtype=np.uint8)
image[:10, :] = 255
image[-10:, :] = 255
image[:, :10] = 255
image[:, -10:] = 255
axes[4].imshow(image, cmap="gray")
axes[4].set_title("Border")

# Remove axes
for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Create image
image = np.array([
    [0, 50, 100],
    [150, 200, 250],
    [255, 128, 64]
])

print("Original image:")
print(image)

print("\nShape:", image.shape)
print("Data type:", image.dtype)
print("Minimum:", image.min())
print("Maximum:", image.max())
print("Mean:", image.mean())

# 1. Access three individual pixels
print("Three pixels:")
print(image[0, 1])
print(image[1, 2])
print(image[2, 0])

# 2. Change the center pixel
image[1, 1] = 255

# 3. Change the first row
image[0, :] = 255

# 4. Change the last column
image[:, 2] = 0

print("\nModified image:")
print(image)

# 5. Print statistics
print("\nAfter modifications:")
print("Minimum:", image.min())
print("Maximum:", image.max())
print("Mean:", image.mean())
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1. Load image and ensure it's in RGB mode
img = Image.open(r"\\192.168.1.10\shear all\lena5.jpg").convert("RGB")
rgb_array = np.array(img)

# 2. Extract individual Red, Green, and Blue channels
R = rgb_array[:, :, 0]  # Red channel intensity (0-255)
G = rgb_array[:, :, 1]  # Green channel intensity (0-255)
B = rgb_array[:, :, 2]  # Blue channel intensity (0-255)

# 3. Display full RGB image alongside individual color channels
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Full RGB image
axes[0].imshow(rgb_array)
axes[0].set_title("Full RGB Image")
axes[0].axis("off")

# Red Channel
axes[1].imshow(R, cmap="Reds")
axes[1].set_title("Red Channel (R)")
axes[1].axis("off")

# Green Channel
axes[2].imshow(G, cmap="Greens")
axes[2].set_title("Green Channel (G)")
axes[2].axis("off")

# Blue Channel
axes[3].imshow(B, cmap="Blues")
axes[3].set_title("Blue Channel (B)")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# 1. Load an image
image_path = r"\\192.168.1.10\shear all\lena5.jpg"
img = Image.open(image_path)
image_array = np.array(img)

# 2. Display the original image
plt.figure(figsize=(5, 5))
if len(image_array.shape) == 2:
  plt.imshow(image_array, cmap="gray")
else:
  plt.imshow(image_array)
plt.title("Original Image")
plt.axis("off")
plt.show()

# 3. Print width and height
height, width = image_array.shape[:2]
print(f"3. Width: {width}")
print(f"   Height: {height}")

# 4. Print number of channels when applicable
if len(image_array.shape) == 3:
  channels = image_array.shape[2]
  print(f"4. Number of Channels: {channels}")
else:
  print("4. Number of Channels: 1 (Grayscale image)")

# 5. Print data type
print(f"5. Data Type: {image_array.dtype}")

# 6. Print minimum and maximum values
print(f"6. Minimum Value: {image_array.min()}")
print(f"   Maximum Value: {image_array.max()}")

# 7. Calculate average pixel value
avg_value = image_array.mean()
print(f"7. Average Pixel Value: {avg_value:.2f}")

# 8. Display the RGB channels separately when the image is RGB
if len(image_array.shape) == 3 and image_array.shape[2] >= 3:
  R = image_array[:, :, 0]
  G = image_array[:, :, 1]
  B = image_array[:, :, 2]

  fig, axes = plt.subplots(1, 3, figsize=(12, 4))

  axes[0].imshow(R, cmap="Reds")
  axes[0].set_title("Red Channel")
  axes[0].axis("off")

  axes[1].imshow(G, cmap="Greens")
  axes[1].set_title("Green Channel")
  axes[1].axis("off")

  axes[2].imshow(B, cmap="Blues")
  axes[2].set_title("Blue Channel")
  axes[2].axis("off")

  plt.tight_layout()
  plt.show()
else:
  print("8. Image is not RGB; individual channel display skipped.")
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# 1. تحميل الصورة بتدرج الرمادي (Grayscale)
image_path = r"C:\Users\PC-LAB1\Desktop\lena.jpg"  # مثال على مسار سطح المكتب
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# (تنبيه: إذا لم توفر الصورة، يمكنك استخدام مصفوفة وهمية للتجربة)
if img is None:
  print("لم يتم العثور على الصورة، جاري إنشاء صورة افتراضية...")
  img = np.random.randint(50, 120, (400, 400), dtype=np.uint8)

# 2. تطبيق Histogram Equalization لتحسين التباين
equalized_img = cv2.equalizeHist(img)

# 3. عرض الصورة الأصلية والمعدلة مع الهيستوجرام الخاص بكل منهما
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# الصورة الأصلية
axes[0, 0].imshow(img, cmap="gray")
axes[0, 0].set_title("Original Image")
axes[0, 0].axis("off")

# هيستوجرام الصورة الأصلية
axes[0, 1].hist(img.ravel(), bins=256, range=[0, 256], color="black")
axes[0, 1].set_title("Original Histogram")
axes[0, 1].set_xlim([0, 256])

# الصورة بعد التعديل (Equalized)
axes[1, 0].imshow(equalized_img, cmap="gray")
axes[1, 0].set_title("Equalized Image (Better Contrast)")
axes[1, 0].axis("off")

# هيستوجرام الصورة المعدلة
axes[1, 1].hist(
    equalized_img.ravel(), bins=256, range=[0, 256], color="black"
)
axes[1, 1].set_title("Equalized Histogram")
axes[1, 1].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```
